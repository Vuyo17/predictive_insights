# Labour Market Status Prediction - Hackathon Starter

This notebook gives a quick workflow to run the baseline model and inspect outputs.

```python
from pathlib import Path
import pandas as pd

data_dir = Path('data')
print('Files in data/:')
for p in sorted(data_dir.glob('*')):
    print('-', p.name)
```

```python
import subprocess

cmd = [
    'python',
    'src/train_and_submit.py',
]

result = subprocess.run(cmd, capture_output=True, text=True)
print(result.stdout)
if result.returncode != 0:
    print(result.stderr)
    raise RuntimeError('Training command failed')
```

```python
metrics_path = Path('outputs/cv_metrics.json')
submission_path = Path('outputs/submission.csv')

if metrics_path.exists():
    print(metrics_path.read_text())

if submission_path.exists():
    sub = pd.read_csv(submission_path)
    display(sub.head())
    print('Rows:', len(sub))
    print('Any duplicate IDs:', sub['anonymised_id'].duplicated().any())
    print('Any missing predictions:', sub['employed_status'].isna().any())
```
